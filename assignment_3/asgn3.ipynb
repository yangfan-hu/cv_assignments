```python
# Setup
import numpy as np
import cv2
import matplotlib.pyplot as plt

%matplotlib inline
plt.rcParams['figure.figsize'] = (15.0, 12.0) # set default size of plots
plt.rcParams['image.interpolation'] = 'nearest'
plt.rcParams['image.cmap'] = 'gray'

# for auto-reloading extenrnal modules
%load_ext autoreload
%autoreload 2
```

```python

# 1. 读取图像
img = cv2.imread('truck.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 考虑使用滤波平滑图像以减少噪声对边缘检测的影响
# 仅供参考，自行选择是否使用，以及参数 
gray = cv2.GaussianBlur(gray, (3, 3), 0)

# 2. 使用 OpenCV 进行边缘检测 (Canny)
# 这一步是必须的，Hough 算法处理的是二值化的边缘点
edges = cv2.Canny(gray, 100, 200, apertureSize=3)

plt.figure(figsize=(10, 10))
plt.imshow(edges, cmap='gray')
plt.title("Edge Detection (Canny)")
plt.axis('off')
plt.show()
```

```python
def HoughLines(image, rho_res, theta_res, threshold, min_line_len=50):
    """
    手动实现霍夫直线检测 (返回端点格式)
    image: 边缘检测后的二值图 (Canny)
    rho_res: 径向分辨率 (通常为 1)
    theta_res: 角度分辨率 (通常为 np.pi/180)
    threshold: 累加器阈值
    min_line_len: 线段最小长度 (用于构造端点)
    返回格式对齐 OpenCV: (N, 1, 4) -> [[x1, y1, x2, y2]]
    """
```

```python
# 调用函数

lines = HoughLines(edges, 1, np.pi/180, 170,100)

output = img.copy()

# 绘制直线
if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line[0]
        cv2.line(output, (x1, y1), (x2, y2), (0, 0, 255), 3)

# 显示最终结果
plt.figure(figsize=(10, 10))
plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title("Final Detection Results")
plt.axis('off')
plt.show()
```

```python

# 1. 读取图像
img = cv2.imread('coins.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# 考虑使用滤波平滑图像以减少噪声对边缘检测的影响
# 仅供参考，自行选择是否使用，以及参数 
gray = cv2.GaussianBlur(gray, (5, 5), 0)

# 2. 使用 OpenCV 进行边缘检测 (Canny)
# 这一步是必须的，Hough 算法处理的是二值化的边缘点
edges = cv2.Canny(gray, 50, 150, apertureSize=3)

plt.figure(figsize=(10, 10))
plt.imshow(edges, cmap='gray')
plt.title("Edge Detection (Canny)")
plt.axis('off')
plt.show()
```

```python
def HoughCircles(image, min_radius, max_radius, threshold, center_threshold=10):
    """
    手动实现霍夫圆检测 (类似 OpenCV 逻辑)
    image: 输入灰度图
    min_radius: 最小半径
    max_radius: 最大半径
    threshold: 累加器阈值 (控制灵敏度)
    center_threshold: 抑制重复圆心的距离阈值 (仁慈模式的关键)
    返回格式对齐 OpenCV: (1, N, 3) -> [[x_center, y_center, radius]]
    """
```

```python
# 调用函数 (假设检测半径在 20-50 之间的圆)
circles = HoughCircles(edges, min_radius=20, max_radius=50, threshold=15)

output = img.copy()

# 绘制圆
if circles is not None:
    circles = np.uint16(np.around(circles))
    for i in circles[0, :]:
        # 绘制外圆
        cv2.circle(output, (i[0], i[1]), i[2], (0, 255, 0), 3)
        # 绘制圆心
        cv2.circle(output, (i[0], i[1]), 2, (255, 0, 0), 3)

# 显示最终结果
plt.figure(figsize=(10, 10))
plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title("Final Detection Results")
plt.axis('off')
plt.show()
```
